```python
def read_csv_auto_encoding(filepath):
    # Detectar a codificação
    with open(filepath, 'rb') as f:
        result = chardet.detect(f.read(10000))  # Lê os primeiros 10.000 bytes
        encoding = result['encoding']
        print(f"Codificação detectada: {encoding}")
        
        # Ler o CSV com a codificação detectada
    df = pd.read_csv(filepath, encoding=encoding)
    return df
```

## Funções

```python
# Função para calcular idade
def calcular_idade(data_nasc):
    hoje = datetime.today()
    idade = hoje.year - data_nasc.year - ((hoje.month, hoje.day) < (data_nasc.month, data_nasc.day))
    return idade
```

```python
df["age"] = df["birth_date"].apply(calcular_idade)
```

## Automação e Pipelines

```python
import pandas as pd
import os

def read_data(filename):
    path = r"C:\Users\USUARIO\Documents\Git-arquivos\DataWrangling_Cleaning"
    return pd.read_csv(os.path.join(path, filename))
    
    
def basic_info(df):
    print("Formato do dataset (linhas, colunas):", df.shape)
    missing_values = df.isnull().sum()
    missing_values = missing_values[missing_values > 0]
    if missing_values.empty:
        print("Base sem dados faltantes")
    else:
        print("Base com dados faltantes")
        print("Colunas com dados faltantes", missing_values.index)
        

def encoding_column(df,column_name):
    df_dummies_features = pd.get_dummies(df[column_name], dtype=int, prefix=column_name)
    return pd.concat([df.drop(column_name, axis=1), df_dummies_features], axis=1)
    

def transform_category_columns(df):
    columns = df.select_dtypes(include="object").columns
    for col in columns:
        df = encoding_column(df, col)
    return df
    

def normalize_column(df, column_name):
    df[column_name] = (df[column_name] - df[column_name].min()) / (df[column_name].max() - df[column_name].min())
    return df
    

def normalize_numeric_columns(df):
    columns = df.select_dtypes(include="number").columns
    for col in columns:
        df = normalize_column(df, col)
    return df
    

def process(filename):
    df = read_data(filename)
    basic_info(df)
    df = transform_category_columns(df)
    df = normalize_numeric_columns(df)
    return df
```

```python
process("gym_members_exercise_tracking.csv")
```
